```python
# Remove rows where duration is missing:

movies = df[df['type'] == 'Movie'].copy()

movies = movies.dropna(subset=['duration'])

movies['duration_min'] = movies['duration'].str.replace(' min', '').astype(int)
```

```python
# Find the Longest Movie

longest_movie = movies.loc[movies['duration_min'].idxmax()]

print(longest_movie[['title', 'duration']])
```

```python
# Insight:
The longest movie in the Netflix dataset is Black Mirror: Bandersnatch with a duration of 312 minutes.
```

```python
# Compare Movies vs TV Shows by Country

country_type = pd.crosstab(
    df['country'],
    df['type']
)

print(country_type.head())
```

```python
# Countries That Focus More on TV Shows

country_type = pd.crosstab(
    df['country'],
    df['type']
)

tv_focused = country_type[
    country_type['TV Show'] > country_type['Movie']
]

tv_focused.sort_values(
    by='TV Show',
    ascending=False
).head(10)
```

```python
# Insight:
# The United Kingdom had the highest number of TV Shows (213), slightly exceeding its Movie count (206), making it one of the strongest TV Show-focused countries in the dataset. 
# Japan and South Korea showed an even stronger preference for TV Shows, producing more than twice as many TV Shows as Movies.
```

```python
# Convert Date

df['date_added'] = pd.to_datetime(df['date_added'])
```

```python
# Extract Year Added

df['year_added'] = df['date_added'].dt.year
```

```python
# Split Genres (This creates one row per genre.)

genres_df = df.copy()

genres_df['listed_in'] = genres_df['listed_in'].str.split(', ')

genres_df = genres_df.explode('listed_in')
```

```python
# Count Genres by Year

genre_trends = pd.crosstab(
    genres_df['year_added'],
    genres_df['listed_in']
)

print(genre_trends.tail())
```

```python
# Fastest-Growing Genres

growth = genre_trends.iloc[-1] - genre_trends.iloc[0]

growth.sort_values(ascending=False).head(10)
```

```python
# Visualization

top_growth = growth.sort_values(
    ascending=False
).head(5)

sns.barplot(
    x=top_growth.values,
    y=top_growth.index
)

plt.title("Fastest Growing Netflix Genres")
plt.xlabel("Growth in Number of Titles")
plt.ylabel("Genre")

plt.show()
```

```python
# Insight:
# International Movies and Dramas experienced the strongest growth in Netflix's content library during recent years.
```
